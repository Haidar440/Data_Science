## Feature Engineering For House Pricing

We will be perform the below steps
1. Handling Missing Values
2. Temporal Variables
3. Handle Categorical Variables: remove rare labels
4. Standarddize the values of the variable to the ame range

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
pd.set_option("display.max_columns",None)
```

```python
dataset = pd.read_csv("train.csv")
dataset.head()
```

```python
#missing value
# missing_value = dataset.columns[dataset.isnull().any()]
missing_value = dataset.isnull().sum()
missing_value = missing_value[missing_value>0]
missing_precent = (missing_value/len(dataset))*100
missing_df = pd.DataFrame({
    'missing_value':missing_value,
    'missing %':np.round(missing_precent,2)
})
missing_df.sort_values("missing %",ascending=False)
```

```python
# Replace Categorial Feature 
# categorial_features = [feature for feature in dataset.columns if dataset[feature].dtype=="O"]
categorial_features = dataset.select_dtypes(include="O").columns
categorial_features
```

```python
data = dataset.copy()
def replace_cat_feature(dataset,categorial_features):
    data[categorial_features] =  data[categorial_features].fillna(data[categorial_features].mode().iloc[0])
    return data
```

```python
replace_cat_feature(data,categorial_features)
# data["Alley"].fillna(data["Alley"].mode()[0])
# data['Alley'].mode()
```

```python
# numerical columns with missing values
# numerical_cols = dataset.select_dtypes(exclude='O')
# numerical_nan_cols = numerical_cols.columns[numerical_cols.isnull().any()]
# numerical_nan_cols #mine approch
numerical_feature_nan = [feature for feature in dataset.columns if dataset[feature].dtype!="O" and dataset[feature].isnull().sum()>0] #tutor Approch
numerical_feature_nan
```

```python
import warnings

# Ignore all warnings
warnings.filterwarnings("ignore")
for feature in numerical_feature_nan:
    median_values = dataset[feature].median()
    dataset[feature+'NaN'] = np.where(dataset[feature].isnull(),1,0)
    dataset[feature].fillna(median_values,inplace=True)
```

```python
dataset[numerical_feature_nan].isnull().sum()
```

```python
# temporal Variable(eg,Datatime, time)
temporal_variable = [feature for feature in dataset.columns if ("Yr" in feature or "Year" in feature) and "NaN" not in feature]
temporal_variable.remove("YrSold")
for feature in temporal_variable:
    dataset[feature+'age of house'] = dataset["YrSold"]-dataset[feature] 
```
