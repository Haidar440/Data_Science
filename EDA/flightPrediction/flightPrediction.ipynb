```python
#importing basics libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_excel("Data_Train.xlsx")
```

```python
# dataset["Date"] = dataset["Date_of_Journey"].str.split("/").str[0]
# dataset["Month"] = dataset["Date_of_Journey"].str.split("/").str[1]
# dataset['Year'] = dataset['Date_of_Journey'].str.split("/").str[2] 
# dataset[["Date","Month","Year"]] = dataset['Date_of_Journey'].apply(lambda x:pd.Series(x.split("/")))
dataset[["Date","Month","Year"]] = dataset["Date_of_Journey"].str.split('/',expand=True)
# dataset
```

```python
dataset.info()
```

```python
#now convert object to int whatever we splited values
# # dataset["Date"] = dataset['Date'].astype(int)
# dataset[["Date","Month","Year"]] = dataset[["Date","Month","Year"]].apply(lambda x : x.astype(int))
dataset[["Date","Month","Year"]] = dataset[["Date","Month","Year"]].astype(int)
# dataset.info()
```

```python
#now main feature should be drop after split from there
dataset.drop(columns="Date_of_Journey", inplace=True,axis=1)
```

```python
dataset['Arrival_Time'] = dataset["Arrival_Time"].str.split(" ").str[0]
dataset['Arrival_Time']
```

```python
dataset[["Arrival_hour","Arrival_minutes"]] = dataset['Arrival_Time'].str.split(":",expand=True).astype(int)
dataset.drop(columns="Arrival_Time")
dataset.info()
```

```python
# dataset[["Dep_hour","Dep_min"]] = dataset['Dep_Time'].apply(lambda x:pd.Series(x.split(":"))).astype(int)
dataset[["Dep_hour","Dep_min"]] = dataset['Dep_Time'].str.split(":",expand=True).astype(int)
dataset.drop("Dep_Time",inplace=True,axis=1)
```
